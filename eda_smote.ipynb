```python
#Contamos cuántas imágenes hay por clase
conteo_clases = data['Clase'].value_counts().sort_index()

#Definimos el orden de las clases
orden_clases = [1, 2, 3]  # 1: Baja

# Fill NaN values with 0 to avoid the conversion error
conteo_clases = conteo_clases.fillna(0)

#Creamos el gráfico de barras
plt.figure(figsize=(10,6))
bars = plt.bar(conteo_clases.index.astype(str), conteo_clases.values, color=['#4CAF50', '#FF9800', '#F44336'])

#Personalizamos el gráfico
plt.title("Cantidad Total de Imágenes por Clase de Patología Ovárica")
plt.xlabel("Tipo de Patología")
plt.ylabel("Número de Imágenes")
plt.xticks(rotation=25, ha='right')

#Agregamos etiquetas con los valores encima de cada barra
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, height + 5,
             f'{int(height)}', ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.show()
```

```python
# Imprimir cantidad de imágenes por clase
print("Cantidad de imágenes por clase:")
print(conteo_clases)
```

```python
# Aplicar técnicas de over-sampling en datos con pocas muestras

from imblearn.over_sampling import SMOTE
smote = SMOTE(random_state=42)
X = data.drop('Clase', axis=1)
y = data['Clase']
# Por ejemplo, todas al menos a 100 muestras
original_first = int(conteo_clases[1] * 1.5)
original_third = int(conteo_clases[3] * 4)

sampling_strategy = {
    1: original_first,
    2: 735,
    3: original_third,
}
"""sampling_strategy = {
    1: 280,
    2: 735,
    3: 120,
}"""
smote = SMOTE(sampling_strategy=sampling_strategy, random_state=42)
# Aplicar SMOTE
X_resampled, y_resampled = smote.fit_resample(X, y)
data_resampled = pd.concat([X_resampled, y_resampled], axis=1)
data_resampled
# Lista de correlaciones para la columna 'Clase'
conteo_clases = data_resampled['Clase'].value_counts()
print("Cantidad de imágenes por clase:")
print(conteo_clases)

# Guardar el dataset balanceado
data_resampled.to_csv("Dataset_train_balanceado.csv", index=False)
```

```python
#Después, se realizó una matriz de correlación para visualizar las relaciones numéricas entre las variables radiómicas.
#Primero se convierte la Clase a valores numéricos para poder incluirla en la matriz

#Después, se realizó una matriz de correlación para visualizar las relaciones numéricas entre las variables radiómicas.
#Primero se convierte la Clase a valores numéricos para poder incluirla en la matriz

from sklearn.preprocessing import LabelEncoder
#encoder = LabelEncoder()
#data['Clase_Num'] = encoder.fit_transform(data['Clase'])

# Check the actual column names in your dataframe
# print(data.columns)

# Create the correlation matrix with the columns that actually exist in your dataframe
# Removed 'IntensidadMedia' and 'DesvStd' which don't exist in the dataframe
"""matriz_de_correlación = data[['Clase_Num','area',
                              'skewness_val', 'kurtosis_val', 'contraste', 'correlacion',
                              'homogeneidad', 'energia', 'entropia', 'lbp_mean',
                              'perimetro', 'circularidad', 'compacidad']].corr()"""

nombres_en ={
    'intensidad_media': 'mean_intensity',
    'mediana': 'median',
    'desviacion_std': 'std_dev',
    'varianza': 'variance',
    'asimetria': 'skewness',
    'curtosis': 'kurtosis',
    'p10': 'p10',
    'p90': 'p90',
    'rango': 'range',
    'homogeneidad': 'homogeneity',
    'energia': 'energy',
    'contraste': 'contrast',
    'correlacion': 'correlation',
    'entropia': 'entropy',
    'lbp_entropia': 'lbp_entropy',
    'lbp_energia': 'lbp_energy',
    'area': 'area',
    'circularidad': 'circularity',
    'compacidad': 'compactness',
    'solidez': 'solidity',
    'excentricidad': 'eccentricity',
    'extension': 'extent',
    'Clase': 'Class'
}

# Fixed: Use 'data' instead of undefined 'df'
df = data.rename(columns=nombres_en)
corr_matrix = df.corr()

plt.figure(figsize=(18,10))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title("Pearson Correlation Heatmap between Class and Radiomic Features")
plt.tight_layout()
#plt.savefig("Figura3_CorrelationHeatmap.pdf", bbox_inches='tight')
plt.show()

```

```python
target = "Class"

corr_target = corr_matrix[target].drop(target).sort_values(ascending=False)

top_pos = corr_target.head(4)
top_neg = corr_target.tail(4).sort_values()

features = list(top_pos.index) + list(top_neg.index)
sub = corr_matrix.loc[[target] + features, [target] + features]

plt.figure(figsize=(10, 8))
sns.heatmap(sub, annot=True, fmt=".2f",
            cmap="coolwarm", center=0, vmin=-1, vmax=1)
plt.title("Pearson Correlation Heatmap between Class and the most correlated radiomic features")
plt.tight_layout()
#plt.savefig("Figura3_CorrelationHeatmap.pdf", bbox_inches="tight")
plt.show()
```

```python
#Procedemos a cambiar los valores de la columna "Clase" por sus respectivos nombres de patologías
data['Clase'] = data['Clase'].replace({
    1: 'Benigno',
    2: 'Borderline',
    3: 'Maligno',
    })
# First, define the 'clases' variable by extracting unique values from the 'Clase' column
clases = data['Clase'].unique()  # This creates an array of unique class values

# Now the rest of your code will work
for c in clases:
    subset = data[data['Clase'] == c]
    matriz_corr = subset[['area', 'correlacion','homogeneidad','compacidad','energia','lbp_entropia','varianza','rango','desviacion_std','extension','p90','entropia','curtosis','mediana','intensidad_media','asimetria','p10','lbp_energia','circularidad','excentricidad','solidez','contraste']].corr()

    plt.figure(figsize=(14,10))
    sns.heatmap(matriz_corr, annot=True, cmap='coolwarm')
    plt.title(f"Mapa de correlación - Clase {c}")
    plt.tight_layout()
    plt.show()
```
